```python
import torch
import torch.nn as nn
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
#exercise to demonstrate and view the performance/results of a linear regression problem solved using 
#OLS method and building a neural network
```

```python
df = pd.read_csv('Datasets/Experience-Salary.csv')
```

```python
X=df.iloc[:,0:1].values
y=df.iloc[:,1].values

X = torch.tensor(X)
y = torch.tensor(y)

print(X.shape,'\n',y.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
lR = LinearRegression()
lR.fit(X_train, y_train)
```

```python
y_test_pred = lR.predict(X_test)
```

```python
((y_test_pred - y_test.numpy())**2).sum()/y_test.shape[0]
```

```python
print('OLS method coefficients, m and c:',lR.coef_[0], lR.intercept_)
```

```python
#build a neural network and predict the salary with experience as the alone feature
```

```python
class RegressionNetwork(torch.nn.Module):
    def __init__(self):
        super(RegressionNetwork, self).__init__()
        self.m = nn.Parameter(torch.tensor(0.0))
        self.c = nn.Parameter(torch.tensor(0.0))
    def forward(self, x):
        y = self.m*x + self.c
        return y
```

```python
network = RegressionNetwork()
optimizer = torch.optim.SGD(network.parameters(), lr=0.001)
```

```python
X = torch.tensor(X, dtype=float64)
y = torch.tensor(y, dtype=float64)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
epoch = 100

for i in range(100):
    y_pred = network(X_train)
    y_pred = y_pred.reshape(-1)
    loss = ((y_pred - y_train)**2).mean()
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (i%10 == 0):
        print(i, loss, network.m.grad, network.c.grad)
```

```python
y_pred_test = network(X_test)
y_pred_test = y_pred_test.reshape(-1)
```

```python
loss = ((y_pred_test - y_test)**2).mean()
```
