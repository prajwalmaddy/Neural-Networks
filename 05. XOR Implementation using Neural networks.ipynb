```python
import numpy as np
import torch
import torch.nn as nn
```

```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super(NeuralNetwork, self).__init__()
        self.hidden1 = nn.Linear(2,1, dtype=torch.float64)
        self.hidden2 = nn.Linear(2,1, dtype=torch.float64)
        self.out = nn.Linear(2,1, dtype=torch.float64)

    def forward(self,x):
        h1 = torch.sigmoid(self.hidden1(x))
        h2 = torch.sigmoid(self.hidden2(x))
        y = torch.sigmoid(self.out(torch.cat([h1,h2],dim=1)))

        return y
```

```python
X = torch.tensor([[0,0], [0,1], [1,0], [1,1]], dtype=torch.float64)
y = torch.tensor([0,1,1,0], dtype=torch.float64).reshape(-1,1)
```

```python
XOR_classifier = NeuralNetwork()
epoch = 10000
optimizer = torch.optim.SGD(XOR_classifier.parameters(), lr=0.1)
```

```python
y_pred = XOR_classifier(X)
```
