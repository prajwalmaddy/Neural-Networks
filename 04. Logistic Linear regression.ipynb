```python
import torch
import torch.nn as nn
from sklearn.preprocessing import StandardScaler
```

```python
#normalize the data
scaler = StandardScaler()
```

```python
X = X.values
y = y.values
```

```python
X = scaler.fit_transform(X)
#y = scaler.fit_transform(y)
```

```python
X = torch.tensor(X.astype(np.float32))
y = torch.tensor(y, dtype=torch.float32).reshape(-1,1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super(NeuralNetwork, self).__init__()
        #self.m = nn.Parameter(torch.zeros(3,1)) # set of weights
        #self.c = nn.Parameter(torch.zeros(1,1)) # 1 bias in the form of a tensor
        self.linear = nn.Linear(3,1)
        
    def forward(self, x):
        #y = (x@self.m + self.c)
        y = self.linear(x)
        return y

    # first draft: I am not including the sigmoid function here as part of the learning. Because- let the algorithm learn
    # and sigmoid operation can be included later
```

```python
regression = NeuralNetwork()
epoch = 400
optimizer = torch.optim.SGD(regression.parameters(), lr=0.01)
```

```python
for i in range(epoch):
    y_pred = regression(X_train)
    #y_prob = torch.sigmoid(y_pred)

    criterion = nn.BCEWithLogitsLoss()
    loss = criterion(y_pred, y_train.float())

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (i%50 == 0):
        #print(i, loss, regression.m.grad, regression.c.grad)
        print(i, loss)
```

```python
y_pred = regression(X_test)
y_pred_prob = torch.sigmoid(y_pred)
```

```python
y_res = torch.where(y_pred_prob >= 0.5, 1.0, 0.0)
```

```python
y_res.numpy().reshape(1,-1)
```

```python
y_test.numpy().reshape(1,-1)
```

```python
print("Metrics using Neural networks:")
print("Accuracy", metrics.accuracy_score(y_res, y_test))
print("F1:",metrics.f1_score(y_res, y_test))
print("Confusion matrix:", metrics.confusion_matrix(y_res, y_test))
```
