```python
import numpy as np
import torch
import torch.nn as nn
```

```python
class NeuralNetwork(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size

        self.weights_input_hidden = torch.random(self.input_size, self.hidden_size)
        self.weights_hidden_output = torch.random(self.hidden_size, self.output_size)

        self.bias_hidden = torch.zeros(1, self.hidden_size)
        self.bias_output = torch.zeros(1, self.output_size)

        self.sigmoid = nn.sigmoid()


    def feedforward(self, X):
         self.hidden_activation = X@self.weights_input_hidden + self.bias_hidden
         self.hidden_output = self.sigmoid(self.hidden_activation)

    
```
