## Necessary libraries

```python
import torch
import torch.nn as nn
import numpy as np
```

## Defining the forward pass of the architecture

```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super(NeuralNetwork, self).__init__()
        self.hidden = nn.Linear(2,4, dtype=torch.float32)
        self.output = nn.Linear(4,1, dtype=torch.float32)
        self.actvn = nn.ReLU()
        
    def forward(self,x):
        x = self.hidden(x)
        x = self.actvn(x)
        
        out = self.output(x)
        return out
```

## XOR Truth table

```python
X = torch.tensor([[0,0], [0,1], [1,0], [1,1]], dtype=torch.float32)
y = torch.tensor([0,1,1,0], dtype=torch.float32).reshape(-1,1)
```

## Network and optimizers definition

```python
XOR_classifier = NeuralNetwork()
epoch = 8000
optimizer = torch.optim.SGD(XOR_classifier.parameters(), lr=0.01)
```

## Backward propagation with loss calculation

```python
for i in range(epoch):
    y_pred = XOR_classifier(X)

    criterion = nn.BCEWithLogitsLoss()
    loss = criterion(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (i%500 == 0):
        #print(i, loss, regression.m.grad, regression.c.grad)
        print(i, loss)
```
